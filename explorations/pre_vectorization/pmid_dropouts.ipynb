# Dropout of scientific papers
The following graph makes it visual at which step papers are slipping through our filter, giving insights where the pipeline can gain meaningful improvements.
(Sankey diagram)

## Approach with Plotly

```python
# imports
import plotly.graph_objects as go
```

```python
# values TODO: Change values below for figure in thesis
v_input = 935
v_a_openalex = 933
v_u_openalex = v_input - v_a_openalex
v_a_openaccess = 359
v_u_openaccess = v_a_openalex - v_a_openaccess
v_a_pdf_url = 240
v_u_pdf_url = v_a_openaccess - v_a_pdf_url
v_s_download = 168
v_u_download = v_a_pdf_url - v_s_download
v_s_parse = 152
v_u_parse = v_s_download - v_s_parse
```

```python
# in list
l_values = [
    v_input,
    v_a_openalex,
    v_u_openalex,
    v_a_openaccess,
    v_u_openaccess,
    v_a_pdf_url,
    v_u_pdf_url,
    v_s_download,
    v_u_download,
    v_s_parse,
    v_u_parse
]

# node name
nodes = [
    "Input",

    "Available in OpenAlex",
    "Unavailable in OpenAlex",
   
    "Open access",
    "Closed access",

    "PDF URL available",
    "No PDF URL known",

    "Successful PDF download",
    "Unsuccessful PDF download",
    
    "Successful PDF parse",
    "Unsuccessful PDF parse"
]

# zip into a dict for clarity later on
nodes_dict = dict(zip(nodes, list(zip(range(len(nodes)), l_values))))

# add values to displayed labels
labels = [f'{k} | {round(nodes_dict[k][1] / nodes_dict["Input"][1] * 100, 2)}% ({nodes_dict[k][1]})' for k in nodes_dict.keys()]

# colors
colors = [
    "blue",

    "blue",
    "red",
   
    "blue",
    "red",

    "blue",
    "red",

    "blue",
    "red",
    
    "blue",
    "red",
]
```

```python
# actual plotting
fig = go.Figure(data=[go.Sankey(
    arrangement = 'perpendicular',
    node = dict(
        label = labels,
        align='left',
        pad = 25,
        thickness = 20,
        # line = dict(color = "blue", width = 0.5),
        # color = node_colors #"blue",
        # color = "blue",
        color = colors
    ),
    link = dict(
        source = [
            nodes_dict["Input"][0], nodes_dict["Input"][0],
            nodes_dict["Available in OpenAlex"][0], nodes_dict["Available in OpenAlex"][0],
            nodes_dict["Open access"][0], nodes_dict["Open access"][0],
            nodes_dict["PDF URL available"][0], nodes_dict["PDF URL available"][0],
            nodes_dict["Successful PDF download"][0], nodes_dict["Successful PDF download"][0]
        ],

        target = [
            nodes_dict["Available in OpenAlex"][0], nodes_dict["Unavailable in OpenAlex"][0],
            nodes_dict["Open access"][0], nodes_dict["Closed access"][0],
            nodes_dict["PDF URL available"][0], nodes_dict["No PDF URL known"][0],
            nodes_dict["Successful PDF download"][0], nodes_dict["Unsuccessful PDF download"][0],
            nodes_dict["Successful PDF parse"][0], nodes_dict["Unsuccessful PDF parse"][0],
        ],

        value = [
            nodes_dict["Available in OpenAlex"][1], nodes_dict["Unavailable in OpenAlex"][1],
            nodes_dict["Open access"][1], nodes_dict["Closed access"][1],
            nodes_dict["PDF URL available"][1], nodes_dict["No PDF URL known"][1],
            nodes_dict["Successful PDF download"][1], nodes_dict["Unsuccessful PDF download"][1],
            nodes_dict["Successful PDF parse"][1], nodes_dict["Unsuccessful PDF parse"][1],
        ],

        # color = colors
    )
)])

fig.update_layout(title_text="Scientific literature yield", font_size=10)
fig.show()
```
